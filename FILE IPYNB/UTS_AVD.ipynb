### **Outliers Values**

```python
df_temp = df.replace(['ERROR', 'UNKNOWN'], np.nan)
df_temp['Quantity'] = pd.to_numeric(df_temp['Quantity'], errors='coerce')
df_temp['Price Per Unit'] = pd.to_numeric(df_temp['Price Per Unit'], errors='coerce')
df_temp['Total Spent'] = pd.to_numeric(df_temp['Total Spent'], errors='coerce')

results = []
cols = df_temp.select_dtypes(include=['float64', 'int64'])

for col in cols:
    q1 = df_temp[col].quantile(0.25)
    q3 = df_temp[col].quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5*iqr
    upper_bound = q3 + 1.5*iqr
    outliers = df_temp[(df_temp[col] < lower_bound) | (df_temp[col] > upper_bound)]
    percent_outliers = (len(outliers)/len(df_temp))*100
    results.append({'Kolom': col, 'Persentase Outliers': percent_outliers})

results_df = pd.DataFrame(results)
results_df.set_index('Kolom', inplace=True)
results_df = results_df.rename_axis(None, axis=0).rename_axis('Kolom', axis=1)
display(results_df)

# Boxplot
fig, axes = plt.subplots(1, 3, figsize=(15, 6))
num_cols = ['Quantity', 'Price Per Unit', 'Total Spent']

for ax, col in zip(axes, num_cols):
    sns.boxplot(y=df_temp[col], ax=ax)
    ax.set_title(f'Boxplot of {col}')
    ax.set_ylabel(col)

plt.tight_layout()
plt.show()
```

Berdasarkan output outliers di atas, kolom **Quantity** dan **Price Per Unit**
tidak ditemukan outlier dengan persentase **0%**. Namun, kolom **Total Spent**
ditemukan memiliki outlier sebesar **2.59%**.

Hal ini juga terlihat pada boxplot, di mana kolom **Quantity** dan **Price Per Unit** tidak menunjukkan titik di luar whisker, sedangkan kolom **Total Spent**
terdapat satu titik di atas garis batas boxplot yang merepresentasikan nilai **25.0**. Kondisi
ini mengindikasikan adanya transaksi dengan nilai pengeluaran yang relatif tinggi dibandingkan mayoritas transaksi lainnya, sehingga perlu diidentifikasi lebih
lanjut pada tahap **Data Preparation**.

# **Data Preparation**

## **Penanganan Tipe Data**

**1. Quantity**

Kolom Quantity seharusnya bertipe numerik (int) karena berisi jumlah item yang dibeli. Namun sebelum diubah ke tipe integer, nilai tidak valid seperti **ERROR** dan **UNKNOWN** perlu diganti menjadi NaN terlebih dahulu agar proses konversi tidak menghasilkan error.

```python
# Mengganti nilai tidak valid menjadi NaN
df['Quantity'] = df['Quantity'].replace(['ERROR', 'UNKNOWN'], np.nan)

# Diubah ke numerik dan integer
df['Quantity'] = pd.to_numeric(df['Quantity'], errors='coerce')
df['Quantity'] = df['Quantity'].astype('Int64')

df.dtypes
```

**2. Price Per Unit**

Kolom Price Per Unit seharusnya bertipe numerik (float) karena merepresentasikan harga per unit produk. Nilai tidak valid seperti **ERROR** dan **UNKNOWN** perlu diganti menjadi NaN terlebih dahulu sebelum dikonversi ke tipe float.

```python
# Mengganti nilai tidak valid menjadi NaN
df['Price Per Unit'] = df['Price Per Unit'].replace(['ERROR', 'UNKNOWN'], np.nan)

# Diubah ke numerik dan float
df['Price Per Unit'] = pd.to_numeric(df['Price Per Unit'], errors='coerce')
df['Price Per Unit'] = df['Price Per Unit'].astype('Float64')

df.dtypes
```

**3. Total Spent**

Kolom Total Spent seharusnya bertipe numerik (float) karena merepresentasikan total pengeluaran per transaksi. Sama seperti kolom sebelumnya, nilai tidak valid perlu diganti menjadi **NaN** terlebih dahulu sebelum dikonversi.

```python
# Mengganti nilai tidak valid menjadi NaN
df['Total Spent'] = df['Total Spent'].replace(['ERROR', 'UNKNOWN'], np.nan)

# Diubah ke numerik dan float
df['Total Spent'] = pd.to_numeric(df['Total Spent'], errors='coerce')
df['Total Spent'] = df['Total Spent'].astype('Float64')

df.dtypes
```

**4. Transaction Date**

Kolom Transaction Date seharusnya bertipe datetime karena berisi tanggal transaksi. Nilai tidak valid seperti **ERROR** dan **UNKNOWN** perlu diganti menjadi NaN terlebih dahulu sebelum dikonversi ke tipe **datetime**.

```python
# Mengganti nilai tidak valid menjadi NaN
df['Transaction Date'] = df['Transaction Date'].replace(['ERROR', 'UNKNOWN'], np.nan)

# Diubah ke datetime
df['Transaction Date'] = pd.to_datetime(df['Transaction Date'], errors='coerce')

df.dtypes
```

**2. Quantity**

Kolom Quantity memiliki missing values sebesar **4.79%**. Imputasi menggunakan **median** dipilih karena kolom ini bertipe numerik dan median tidak terpengaruh oleh nilai ekstrem sehingga lebih representatif sebagai nilai pengganti dibandingkan mean.

```python
# Before
print("Before")
df[df['Quantity'].isna()]
```

```python
# Imputasi dengan median
pd.options.mode.chained_assignment = None
df['Quantity'] = df['Quantity'].fillna(df['Quantity'].median())

# After
print("After")
df[df['Quantity'].isna()]
```

**3. Item**

Kolom Item memiliki missing values sebesar **9.69%**. Imputasi dilakukan dengan pendekatan berbasis relasi antar kolom, artinya nilai yang kosong diisi berdasarkan nilai **Price Per Unit** yang tersedia pada baris tersebut. Untuk harga yang unik per produk (1.0, 1.5, 2.0, 5.0), Item dapat diisi secara akurat sesuai produk yang
sesuai. Sementara untuk harga yang ambigu (3.0 dan 4.0) yang masing-masing dimiliki oleh dua produk, Item ditentukan berdasarkan modus per kelompok harga, harga 3.0 diisi dengan Juice karena lebih sering muncul dibandingkan Cake, dan harga 4.0 diisi dengan Sandwich karena lebih sering muncul dibandingkan Smoothie. Baris yang nilai **Price Per Unit-nya** juga NaN diisi menggunakan modus. Pendekatan ini dipilih untuk meminimalkan bias imputasi dengan memanfaatkan informasi yang sudah tersedia di dalam dataset.

```python
# Pengecekan modus per kelompok
print("Harga 3.0 (Juice vs Cake):")
print(df[df['Price Per Unit'] == 3.0]['Item'].value_counts())

print("\nHarga 4.0 (Sandwich vs Smoothie):")
print(df[df['Price Per Unit'] == 4.0]['Item'].value_counts())
```

```python
# Before
print("Before")
df[df['Item'].isna()]
```

```python
price_to_item = {
    1.0: 'Cookie',
    1.5: 'Tea',
    2.0: 'Coffee',
    5.0: 'Salad',
    3.0: 'Juice',
    4.0: 'Sandwich'
}

# Imputasi Item berdasarkan nilai Price Per Unit
df['Item'] = df.apply(
    lambda row: price_to_item.get(row['Price Per Unit'], row['Item'])
    if pd.isna(row['Item']) else row['Item'], axis=1
)

# Sisa yang masih NaN diisi dengan modus
df['Item'] = df['Item'].fillna(df['Item'].mode()[0])

# After
print("After")
df[df['Item'].isna()]
```

```python
df.groupby('Item')['Price Per Unit'].unique()
```

```python
# Before
print("Before")
df[df['Price Per Unit'].isna()]
```

```python
item_to_price = {
    'Cake': 3.0,
    'Coffee': 2.0,
    'Cookie': 1.0,
    'Juice': 3.0,
    'Salad': 5.0,
    'Sandwich': 4.0,
    'Smoothie': 4.0,
    'Tea': 1.5
}

# Imputasi Price Per Unit berdasarkan mapping harga tetap per produk
df['Price Per Unit'] = df.apply(
    lambda row: item_to_price.get(row['Item'], row['Price Per Unit'])
    if pd.isna(row['Price Per Unit']) else row['Price Per Unit'], axis=1
)

# After
print("After")
df[df['Price Per Unit'].isna()]
```

**5. Total Spent**

Kolom Total Spent memiliki missing values sebesar **5.02%**. Imputasi dilakukan khusus pada baris yang bernilai kosong berdasarkan perhitungan nilai **Quantity × Price Per Unit** karena terbukti seluruh nilai Total Spent di dataset konsisten dengan hasil perkalian kedua kolom tersebut. Pendekatan ini dipilih karena lebih
akurat dibandingkan imputasi median, dengan memanfaatkan relasi matematis antar kolom yang sudah tersedia di dalam dataset.

```python
# Before
print("Before")
df[df['Total Spent'].isna()]
```

```python
# Imputasi Total Spent berdasarkan Quantity x Price Per Unit
df['Total Spent'] = df.apply(
    lambda row: row['Quantity'] * row['Price Per Unit']
    if pd.isna(row['Total Spent']) else row['Total Spent'], axis=1
)

# After
print("After")
df[df['Total Spent'].isna()]
```

Kolom Payment Method memiliki missing values sebesar **31.78%**. Imputasi menggunakan **modus** dipilih karena metode pembayaran yang paling sering digunakan pelanggan menjadi nilai pengganti yang paling representatif.

## **Outliers Values**

```python
results = []
cols = df.select_dtypes(include=['float64', 'int64', 'Int64', 'Float64'])

for col in cols:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5*iqr
    upper_bound = q3 + 1.5*iqr
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    percent_outliers = (len(outliers)/len(df))*100
    results.append({'Kolom': col, 'Persentase Outliers': percent_outliers})

results_df = pd.DataFrame(results)
results_df.set_index('Kolom', inplace=True)
results_df = results_df.rename_axis(None, axis=0).rename_axis('Kolom', axis=1)
display(results_df)

# Boxplot
fig, axes = plt.subplots(1, 3, figsize=(15, 6))
num_cols = ['Quantity', 'Price Per Unit', 'Total Spent']

for ax, col in zip(axes, num_cols):
    sns.boxplot(y=df[col], ax=ax)
    ax.set_title(f'Boxplot of {col}')
    ax.set_ylabel(col)

plt.tight_layout()
plt.show()
```

```python
# Pengecekan baris dengan Total Spent di atas batas atas IQR
q1 = df['Total Spent'].quantile(0.25)
q3 = df['Total Spent'].quantile(0.75)
iqr = q3 - q1
upper_bound = q3 + 1.5 * iqr

print(f'Batas atas IQR: {upper_bound}')
print()
df[df['Total Spent'] > upper_bound][['Item', 'Quantity', 'Price Per Unit', 'Total Spent']]
```

Berdasarkan output outliers di atas, kolom **Quantity** dan **Price Per Unit** tidak ditemukan outlier dengan persentase **0%**. Namun, kolom **Total Spent** ditemukan memiliki outlier sebesar **2.73%**.

Hal ini juga terlihat pada boxplot, di mana kolom **Quantity** dan **Price Per Unit** tidak menunjukkan titik di luar garis batas boxplot, sedangkan kolom **Total Spent** terdapat satu titik di atas garis batas boxplot yang merepresentasikan nilai **25.0**.

Setelah dilakukan pengecekan lebih lanjut, seluruh outlier pada kolom **Total Spent** berasal dari satu kombinasi yang sama, yaitu pembelian **5 item Salad** dengan harga **5.0** per unit sehingga menghasilkan Total Spent sebesar **25.0**. Nilai ini hanya selisih **1.0** dari batas atas IQR sebesar **24.0** dan sepenuhnya valid secara logis karena merupakan transaksi yang wajar dalam operasional kafe. Oleh karena itu, outlier pada kolom **Total Spent** tidak memerlukan penanganan lebih lanjut.
